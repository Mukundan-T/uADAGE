```python
from m_adage import Adage, LinkedAE

from pandas import read_csv
import numpy as np
from sklearn.metrics import mean_squared_error
from tensorflow import zeros

import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns

from json import load
from zipfile import ZipFile
from types import SimpleNamespace
```

### Load models

```python
model_filenames = [
  f'../data/ec_U1_model_{i+1}.keras'
  for i in range(50)
]
```

```python
# Load hyperparameters
with open('../data/ec_U1_model_hps.json', 'r') as f:
  optimal_hps = load(f)
```

```python
train_comp = read_csv('../data/ecmg_lcn01_medmean_normal_muk.csv', index_col = 0)
loaded_models = []

for i, file in enumerate(model_filenames):
  # Create autoencoder object
  ae = LinkedAE.LinkedAE(
    input_dim = train_comp.shape[0],
    encoding_dim = optimal_hps['units'],
    act1 = optimal_hps['act1'],
    act2 = optimal_hps['act2'],
    dropout = optimal_hps['dropout'],
    kl1 = optimal_hps['kl1'],
    kl2 = optimal_hps['kl2'],
    init = optimal_hps['init'],
    tied = optimal_hps['tied'],
    u1 = optimal_hps['u1'],
    u2 = optimal_hps['u2'],
    reg_type = 'u1'
  )
  ae(zeros((1, train_comp.shape[0])))

  # Pull model weights
  with ZipFile(file) as z:
    z.extract('model.weights.h5')

  # Load saved weights
  ae.load_weights('model.weights.h5')

  # Delete extracted weights file
  os.remove('model.weights.h5')
  
  # Create ADAGE object
  adage = Adage.Adage(
    autoencoder = ae,
    history = SimpleNamespace(
      history = {
        'loss': np.loadtxt(
                  f'../data/ec_U1_model_loss_{i+1}.csv',
                  delimiter=','
                ).reshape(-1).tolist(),
        'val_loss': np.loadtxt(
                      f'../data/ec_U1_model_val_loss_{i+1}.csv',
                      delimiter=','
                    ).reshape(-1).tolist()
      }
    ),
    train_comp = train_comp
  )

  # Add model to list
  loaded_models.append(adage)
```

```python
len(loaded_models)
```

### Reconstruction Error

```python
x_train = train_comp.values.astype("float64").T
reconstruct_errors = []

for model in loaded_models:
  x_pred = model.autoencoder.predict(x_train, verbose = 0)
  mse = mean_squared_error(x_train.flatten(), x_pred.flatten())
  reconstruct_errors.append(mse)
```

```python
# Plot distribution of reconstruction error across models
ax = sns.histplot(reconstruct_errors)

plt.title("U1 ADAGE - Reconstruction Error Distribution")
plt.xlabel("Mean Squared Error")
plt.ylabel("Number of Models")

ax.xaxis.set_major_locator(ticker.MaxNLocator(nbins = 6))

# Save plot locally
plot_path = '../data/ec_U1_recon_err.png'
plt.savefig(
  plot_path,
  dpi = 300,
  bbox_inches = 'tight'
)

# Save plot in drive
# !rclone copy '{plot_path}' mygdrive:'Mukundan - Senior Project/U1 Models/plots/'
```

### Training Loss Curves

```python
# Create a NumPy 2D array of losses, where each array is for each model
train_losses = np.array([
  model.history.history['loss']
  for model in loaded_models
])

val_losses = np.array([
  model.history.history['val_loss']
  for model in loaded_models
])

n_models, n_epochs = train_losses.shape
epochs = np.arange(n_epochs)
```

```python
# Calculate median loss across models at each epoch
train_median = np.median(train_losses, axis=0)
val_median = np.median(val_losses, axis=0)

# Calculate 25th and 75th percentiles
train_q25, train_q75 = np.percentile(
  train_losses, [25, 75], axis=0
)
val_q25, val_q75 = np.percentile(
  val_losses, [25, 75], axis=0
)

# Calculate 10th and 90th percentiles
train_q10, train_q90 = np.percentile(
  train_losses, [10, 90], axis=0
)
val_q10, val_q90 = np.percentile(
  val_losses, [10, 90], axis=0
)
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

fig.suptitle(
  'E. coli U1 ADAGE Models - Training vs. Validation Loss Distributions',
  fontsize = 16
)
plt.style.use('tableau-colorblind10')

# *** Plot training loss *** #

# Plot 10th - 90th percentile outside
ax.fill_between(
  epochs + 1,
  train_q10,
  train_q90,
  color = 'tab:blue',
  alpha = 0.15,
  label = 'Training: 10th-90th percentile'
)

# Plot 25th - 75th percentile inside
ax.fill_between(
  epochs + 1,
  train_q25,
  train_q75,
  color = 'tab:blue',
  alpha = 0.3,
  label = 'Training: 25th-75th percentile'
)

# Median training loss
ax.plot(
  epochs + 1,
  train_median,
  color = 'tab:blue',
  linewidth = 2,
  label = 'Median training loss'
)

# *** Plot validation loss *** #

# Plot 10th - 90th percentile outside
ax.fill_between(
  epochs + 1,
  val_q10,
  val_q90,
  color = 'tab:orange',
  alpha = 0.15,
  label = '10th-90th percentile'
)

# Plot 25th - 75th percentile inside
ax.fill_between(
  epochs + 1,
  val_q25,
  val_q75,
  color = 'tab:orange',
  alpha = 0.3,
  label = '25th-75th percentile'
)

# Median validation loss
ax.plot(
  epochs + 1,
  val_median,
  color = 'tab:orange',
  linewidth = 2,
  label = 'Median validation loss'
)

ax.set(
  xlabel = 'Epochs',
  ylabel = 'Loss'
)
ax.grid(alpha = 0.3)

ax.legend(
    loc='best',
    fontsize=9
)

ax.set_ylim(0.006, 0.014)
ax.text(
  0.025, 
  0.05,
  'Initial training loss at end of Epoch 1: 0.02840246 (outside displayed range)',
  transform = ax.transAxes,
  fontsize = 9,
  ha = 'left',
  va = 'bottom',
  bbox = dict(
    facecolor = 'white',
    alpha = 0.85,
    edgecolor = 'gray'
  )
)

plt.tight_layout()

# Save figure locally
plot_path = '../data/ec_U1_loss_dist.png'
fig.savefig(
  plot_path,
  dpi = 300,
  bbox_inches = 'tight'
)

# Save plot in drive
# !rclone copy '{plot_path}' mygdrive:'Mukundan - Senior Project/U1 Models/plots/'
```

### Weight Distributions

```python
plot_titles = [
  f'U1 ADAGE - Model {i+1}'
  for i in range(4)
]

fig, ax = plt.subplots(2, 2, figsize = (15, 10))
fig.subplots_adjust(hspace = 0.25, wspace = 0.25)

fig.suptitle(
  'E. coli U1 ADAGE Models - Weight Distributions',
  fontsize = 16,
  y = 1
)
colorblind_colors = sns.color_palette("colorblind")

for i in range(2):
  for j in range(2):
    weights_matrix = np.asarray(loaded_models[2*i + j].autoencoder.get_weights()[0])

    for node in range(optimal_hps['units']):
      weights = weights_matrix[:, node]
      weights = weights[np.isfinite(weights)]

      # Skip empty arrays and constant distributions
      if len(weights) == 0 or np.min(weights) == np.max(weights):
        continue

      # Each hidden node weight distribution
      sns.histplot(
        weights,
        ax = ax[i, j],
        bins = 30,
        color = colorblind_colors[node % len(colorblind_colors)],
        kde = False
      )

    all_weights = weights_matrix.flatten()
    all_weights = all_weights[np.isfinite(all_weights)]
    cutoff = np.std(all_weights) * 2

    ax[i, j].axvline(
      x = cutoff,
      color = 'black',
      linestyle = '--',
      linewidth = 1
    )
    ax[i, j].axvline(
      x = -cutoff,
      color = 'black',
      linestyle = '--',
      linewidth = 1
    )
    ax[i, j].set(title = plot_titles[2*i + j])

# Save figure locally
plot_path = '../data/ec_U1_weight_dist.png'
fig.savefig(
  plot_path,
  dpi = 300,
  bbox_inches = 'tight'
)

# Save plot in drive
# !rclone copy '{plot_path}' mygdrive:'Mukundan - Senior Project/U1 Models/plots/'
```
