Inspect the KEGG CSV generated from the KEGG database to see how many genes are in each pathway.

### Load libraries

```python
import pandas as pd
```

### Load and filter KEGG TSV

```python
kegg_paths = pd.read_csv(
  '../data/ec_kegg_pathways.txt',
  sep = '\t')
kegg_paths.shape
```

```python
kegg_paths.head()
```

```python
sorted_kegg = kegg_paths.sort_values(by = 'Length')
sorted_kegg.head(130).tail()
```

```python
filtered_kegg = sorted_kegg.iloc[128:]
filtered_kegg.shape
```

```python
filtered_kegg.sort_values(by = 'Length').head()
```

```python
filtered_kegg.to_csv(
  '../data/filtered_ec_kegg_pathways.tsv',
  sep = '\t',
  index = False
)
```

```python
loaded_kegg = pd.read_csv(
  '../data/filtered_ec_kegg_pathways.tsv',
  sep = '\t'
)
loaded_kegg.head()
```
