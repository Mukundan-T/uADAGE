Here we explore the results of the model tuner (top 90 combinations) and save the optimal combination to Google Drive for persistent storage.

### Load libraries

```python
from json import load, dump
```

### Inspect baseline tuner results

Choose the optimal combination based on a higher number of hidden nodes, higher batch size, and same encoding/decoding activation functions (if possible).

```python
# Load tuner results
with open('../data/unreg_tuner_results.json', 'r') as f:
  hps_list = load(f)
```

```python
# Choose best combination
hps = hps_list[1]
hps
```

```python
# Save optimal combination locally
hps_path = '../data/ec_unreg_model_hps.json'
optimal_hps = {param: val for param, val in hps.items() if not param.startswith('tuner')}

with open(hps_path, "w") as f:
  dump(optimal_hps, f)
```

```python
# Save hyperparameter JSON to drive
!rclone copy '{hps_path}' mygdrive:'Mukundan - Senior Project/Unregularized Models/'
```
